```python
import json
```

```python
f=open("Xe_100_10bar.json")
data=json.load(f)
```

```python
import matplotlib.pyplot as plt
print(data.keys())
plt.figure(figsize=(8,15))

plt.subplot(3,1,1)
plt.title("Transverse Diffusion")
plt.plot(data["electric_field"],data["electron_transversal_diffusion"])
plt.ylabel("DT (sqrt(cm))")
plt.subplot(3,1,2)
plt.title("Longitutional Diffusion")
plt.plot(data["electric_field"],data["electron_longitudinal_diffusion"])
plt.ylabel("DL (sqrt(cm))")
plt.subplot(3,1,3)
plt.title("Drift Velocity")
plt.plot(data["electric_field"],data["electron_drift_velocity"])
plt.xlabel("EField (V/cm)")
plt.ylabel("Drift Velocity (cm/ns)")
plt.show
```
